#### Задание 1.1: Метрики MAE и MASE (0,8 балла)

Для оценки качества прогнозирования будем использовать MAE и MASE. Мы уже познакомились с ними на занятиях, а теперь давайте реализуем обе метрики сами (пользоваться готовыми реализациями нельзя!). А при реализации MASE давайте переиспользуем вашу функцию MAE.

$$
\operatorname{MAE} =
\frac{1}{H}\sum_{h=1}^{H}
\left|y_{T+h}-\widehat{y}_{T+h\mid T}\right|
$$

$$
\operatorname{MASE} = \frac{\operatorname{MAE}_{\text{прогноз на test}}}{\operatorname{MAE}_{\text{сезонный наивный прогноз на train}}}
$$

То же самое можно записать так:

$$
\operatorname{MASE} =
\frac{
\frac{1}{H}\sum_{h=1}^{H}\left|y_{T+h}-\widehat{y}_{T+h\mid T}\right|
}{
\frac{1}{T-m}\sum_{t=m+1}^{T}\left|y_t-y_{t-m}\right|
}
$$

$T$ — длина train, $H$ — горизонт прогноза, $m$ — период сезонности (параметр `seasonal_period`). В знаменателе MASE сравниваем значения train со значениями на один сезон раньше. Для несезонного ряда можно оставить `seasonal_period=1`.

Если MAE сезонного наивного прогноза на train равен 0, MASE посчитать не получится. Давайте в этом случае возвращать `np.nan`.
